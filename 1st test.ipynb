```python
from google import genai

# Create Gemini client
client = genai.Client(api_key="ENTER YOUR API KEY HERE")

# Generate response
response = client.models.generate_content(
    #model = genai.GenerativeModel("gemini-3.7-flash")
    model="gemini-3.5-flash-lite",
    contents="What is Model Context Protocol? Explain it to a school student."
)

# Print response
print(response.text)
```

```python
import gradio as gr
from google import genai

client = genai.Client(api_key="ENTER YOUR API KEY HERE")

def generate_text(prompt):
    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )
    return response.text

gr.Interface(
    fn=generate_text,
    inputs=gr.Textbox(label="Enter your prompt"),
    outputs=gr.Textbox(label="Generated Text"),
    title="Gemini AI - PROMPT TO TEXT GENERATOR",
    description="Enter a prompt and generate text using Gemini AI"
).launch()
```
